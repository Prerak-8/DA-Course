```python
# Task 1

import pandas as pd

data = pd.DataFrame({
    "Color": ["Red", "Blue", "Red", "Green", "Blue"]
})

data["Color"] = data["Color"].astype("category")
print(data)
print(data.dtypes)
```

```python
# Task 2

print("Number of unique categories:", len(data["Color"].unique()))

print("Frequency of each category:")
print(data["Color"].value_counts())
```

```python
# Task 3

data_size = pd.DataFrame({
    "Size": ["Small", "Medium", "Large"]
})

data_size["Size"] = pd.Categorical(data_size["Size"], categories = ["Small", "Medium", "Large"], ordered = True)

sorted_size = data_size.sort_values("Size")

print(sorted_size)
```

```python
# Task 4

data_size["Size"] = data_size["Size"].cat.rename_categories({"Small": "S", "Medium": "M", "Large": "L"})

print(data_size)
```

```python
# Challenge

data_size["Size"].value_counts().plot(kind="bar")
```
