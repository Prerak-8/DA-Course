```python
# Task 5

import numpy as np

ages = np.random.randint(10, 81, 100)

plt.hist(ages)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Task 6

months = ["January", "February", "March", "April", "May", "June"]

revenue = [5000, 6500, 7000, 8000, 7500, 9000]
profit = [1200, 1600, 1800, 2200, 2000, 2600]

plt.bar(months, revenue, label="Revenue")
plt.plot(months, profit, label="Profit")

plt.title("Revenue and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

```python
# Task 7

products = ["Shoes", "Belt", "Perfume", "Jacket", "Watch"]
sales = [200, 120, 299, 450, 269]

plt.bar(products, sales, color=["red", "blue", "green", "orange", "purple"])

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid(axis="y")

plt.show()
```

```python
# Task 8

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 180, 150, 240, 200, 170]

plt.plot(months, sales)

highest = max(sales)
lowest = min(sales)

highest_month = months[sales.index(highest)]
lowest_month = months[sales.index(lowest)]

plt.annotate("Highest", (highest_month, highest))
plt.annotate("Lowest", (lowest_month, lowest))

plt.title("Monthly Sales")

plt.show()
```
