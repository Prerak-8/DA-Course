```python
# Task 1

import seaborn as sns

import matplotlib.pyplot as plt
```

```python
# Task 2

tips = sns.load_dataset("tips")
```

```python
# Task 3

print(tips.head())
tips.info()
```

```python
# Task 4

sns.histplot(data = tips, x = "total_bill")
plt.show()
```

```python
# Task 5

sns.kdeplot(data = tips, x = "total_bill")
plt.title("KDE Plot of Total Bill")
plt.show()
```

```python
# Task 6

sns.displot(data=tips, x="total_bill", kde=True)

plt.show()
```

```python
# Task 7

sns.boxplot(x = "day", y = "total_bill", data = tips)
plt.show()
```

```python
# Task 8

sns.violinplot(x = "sex", y = "tip", data = tips)

plt.title("Tip Amount by Gender")
plt.show()
```

```python
# Task 9

sns.stripplot(x = "smoker", y = "tip", data = tips)

plt.title("Tips by Smoker Status")
plt.show()
```

```python
# Task 10

sns.scatterplot(x = "total_bill", y = "tip", data = tips)
plt.show()
```

```python
# Task 11

sns.scatterplot(x = "total_bill", y = "tip", data = tips, hue = "smoker", style = "sex", size = "size")
plt.show()
```

```python
# Task 12

sns.lineplot(x = "day", y = "tip", data = tips)
plt.show()
```

```python
# Task 13

correlation = tips[["total_bill", "tip", "size"]].corr()

sns.heatmap(correlation)
plt.show()
```

```python
# Task 14

correlation = tips[["total_bill", "tip", "size"]].corr()

sns.heatmap(correlation, annot = True, cmap = "coolwarm")
plt.show()
```

```python
# Task 15

correlation = tips[["total_bill", "tip", "size"]].corr()

sns.clustermap(correlation, annot = True, cmap = "coolwarm")
plt.show()
```

```python
# Task 16

sns.regplot(x = "tip", y = "total_bill", data = tips)
plt.show()
```

```python
# Task 17

sns.lmplot(x = "total_bill", y = "tip", hue = "smoker", data = tips)
plt.show()
```

```python
# Task 18

sns.set_theme(style="darkgrid")

sns.scatterplot(x = "total_bill", y = "tip", data = tips)
plt.show()
```

```python
# Task 19

sns.set_palette("pastel")

sns.boxplot(x = "day", y = "total_bill", data = tips)
plt.show()
```

```python
# Task 20

sns.set_context("paper")

sns.scatterplot(x = "total_bill", y = "tip", data = tips)
plt.show()
```
