```python
# Task 1

import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]

product_a = "Watch"
sales_a = [10, 12, 6, 21, 9, 5, 19, 22, 5, 14, 8, 11]

product_b = "Frying pan"
sales_b = [19, 22, 5, 14, 8, 11, 15, 18, 16, 22, 11, 25]

product_c = "Guitar"
sales_c = [9, 2, 5, 14, 6, 11, 6, 8, 16, 12, 10, 3]

plt.stackplot(months, sales_a, sales_b, sales_c, labels = [product_a, product_b, product_c], colors = ["gray", "brown", "sienna"])
plt.title("Sales of 3 products")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 2

months = ["January", "February", "March", "April", "May", "June"]

hr = [2000, 2200, 2100, 2400, 2300, 2500]
marketing = [3000, 3200, 3500, 3400, 3700, 3900]
sales = [5000, 5400, 5800, 6000, 6200, 6500]
it = [2500, 2700, 2900, 3000, 3200, 3400]

plt.stackplot(months, hr, marketing, sales, it, labels = ["HR", "Marketing", "Sales", "IT"])

plt.title("Monthly Revenue by Department")
plt.xlabel("Months")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

```python
# Task 3

product = ["Rolex", "Necklace", "Ring"]
sales = [15000, 11000, 8500]
profit = [3400, 2700, 2100]

fig, axs = plt.subplots(1, 2)

axs[0].bar(product, sales)
axs[0].set_title("Bar Plot")
axs[0].set_xlabel("Products")
axs[0].set_ylabel("Sales")

axs[1].plot(product, profit)
axs[1].set_title("Line Plot")
axs[1].set_xlabel("Products")
axs[1].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```

```python
# Task 4

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

city_1 = [26, 28, 31, 34, 37, 39, 34, 31, 29, 26, 23, 19]
city_2 = [23, 25, 29, 32, 33, 34, 32, 31, 26, 23, 21, 15]
city_3 = [22, 25, 27, 31, 35, 36, 32, 33, 27, 23, 21, 20]

plt.plot(months, city_1, label = "Ahmedabad")
plt.plot(months, city_2, label = "Delhi")
plt.plot(months, city_3, label = "Mumbai")

plt.title("Temperature variation")
plt.xlabel("Months")
plt.ylabel("Temp")
plt.legend(loc = "upper left")
plt.show()
```
