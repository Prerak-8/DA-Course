```python
# Task 4.1

fruits = pd.Series(['apple', 'banana', 'cherry', 'date'])

fruits_count = fruits.count()

print("Number of elements in fruits:", fruits_count)
print("Unique values:", fruits.unique())
print("Count of each unique value:\n", fruits.value_counts())
```

```python
# Task 4.2

data = {
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [20, 21, 19],
    'major': ['CS', 'Engineering', 'Math']
}

student_data = pd.DataFrame(data)

print("First 2 rows:")
print(student_data.head(2))

print("\nLast row:")
print(student_data.tail(1))

print("\nIndex:")
print(student_data.index)

print("\nColumn names:")
print(student_data.columns)

print("\nSummary:")
student_data.info()
```

```python
# Task 4.3

data = [{'name': 'David', 'salary': 50000, 'department': 'Sales'},
{'name': 'Eve', 'salary': 60000, 'department': 'Marketing'},
{'name': 'Frank', 'salary': 55000, 'department': 'Sales'}]

employee_data = pd.DataFrame(data)

print("Only name column:\n", employee_data["name"])
print("\nRow at index 1:\n", employee_data.loc[1])
print("\nName and Salary columns:\n", employee_data[["name", "salary"]])

name_dep = employee_data[["name", "department"]]
print("\nFirst two rows of name and department columns:\n", name_dep.head(2))
```
