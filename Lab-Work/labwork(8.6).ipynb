```python
# Task 4.1

data1 = {
    "Employee_id": [100, 101, 102],
    "Name": ["Alex", "Jason", "Walter"]
}

data2 = {
    "Employee_id": [100, 101, 103],
    "Salary": [15000, 17000, 19000]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

print(df1)
print(df2)
```

```python
# Task 6.1

customers = {
    "Customer_ID": [1, 2, 3],
    "Customer_Name": ["Alex", "Jason", "Walter"]
}

orders = {
    "Order_ID": [101, 102, 103, 104],
    "Customer_ID": [1, 2, 1, 3],
    "Order_Date": ["2024-01-10", "2024-02-15", "2024-03-20", "2024-03-25"],
    "TotalOrderValue": [1200, 800, 1500, 900]
}

customers_df = pd.DataFrame(customers)
orders_df = pd.DataFrame(orders)

print(customers_df)
print(orders_df)
```

```python
# Task 6.2

merged_orders = pd.merge(orders_df, customers_df, on = "Customer_ID", how= "inner")

print(merged_orders)
```

```python
# Task 6.3

high_value_customers = merged_orders[merged_orders["TotalOrderValue"] > 1000]

print(high_value_customers)
```

```python
# Task 7.1

merged_orders["Order_Date"] = pd.to_datetime(merged_orders["Order_Date"])

merged_orders["Month"] = merged_orders["Order_Date"].dt.month

monthly_group = merged_orders.groupby("Month")

print(merged_orders)
```

```python
# Task 7.2

monthly_revenue = monthly_group["TotalOrderValue"].sum()

print("Monthly Revenue:")
print(monthly_revenue)
```

```python
# Task 7.3

monthly_revenue.plot(kind="line")
```

```python
# Task 8.1

products = {
    "Product_ID": [1, 2, 3],
    "Product_Name": ["Laptop", "Phone", "Smart T.V"]
}

sales = {
    "Product_ID": [1, 2, 3],
    "Total_Sales": [5000, 4000, 3000]
}

returns = {
    "Product_ID": [1, 2, 3],
    "Total_Returns": [500, 300, 200]
}

products_df = pd.DataFrame(products)
sales_df = pd.DataFrame(sales)
returns_df = pd.DataFrame(returns)

print(products_df)
print(sales_df)
print(returns_df)
```

```python
# Task 8.2

product_sales = pd.merge(products_df, sales_df, on = "Product_ID", how = "inner")

final_data = pd.merge(product_sales, returns_df, on = "Product_ID", how = "inner")

print(final_data)
```

```python
# Task 8.3

final_data["Net_Sales"] = (final_data["Total_Sales"] - final_data["Total_Returns"])

print(final_data[["Product_Name", "Total_Sales", "Total_Returns", "Net_Sales"]])
```

```python
# Task 8.4

final_data.plot(kind="bar", x = "Product_Name", y = "Net_Sales")
```
