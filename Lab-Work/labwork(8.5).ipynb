```python
# Task 5

data = {
    "name": ["Zack", None, "Charlie"],
    "age": [20, None, 22]
}

df_missing = pd.DataFrame(data)

print("Before filling:")
print(df_missing)

df_filled = df_missing.fillna({
    "name": "Unknown",
    "age": df_missing["age"].mean()
})

print("\nAfter filling:")
print(df_filled)
```

```python
# Task 6

df_dropped = df_missing.dropna()

print(df_dropped)
```

```python
# Task 7

data = {
    "name": ["Zack", "Bob", "Bob", "Charlie"],
    "age": [20, 21, 21, 22]
}

df_duplicates = pd.DataFrame(data)

print("Duplicate rows:")
print(df_duplicates.duplicated())

df_duplicates = df_duplicates.drop_duplicates()

print("\nAfter removing duplicates:")
print(df_duplicates)
```

```python
# Task 8

data = {
    "salary": ["50000", "60000", "55000"]
}

df_numeric = pd.DataFrame(data)

try:
    df_numeric["salary"] = df_numeric["salary"].astype(int)
    print(df_numeric)
except ValueError:
    print("Some values could not be converted to numeric.")
```

```python
# Task 9

data = {
    "name": ["Zack", "Bob", "Charlie"],
    "age": ["20", "21", "22"]
}

df_types = pd.DataFrame(data)

print("Before:")
print(df_types.dtypes)

df_types["age"] = df_types["age"].astype(int)

print("\nAfter:")
print(df_types.dtypes)
```

```python
# Task 10

data = {
    "date": ["2026-01-10", "2026-02-15", "2026-03-20"]
}

df_date = pd.DataFrame(data)

df_date["date"] = pd.to_datetime(df_date["date"])

print(df_date)
print(df_date.dtypes)
```

```python
# Task 11

df_date["year"] = df_date["date"].dt.year
df_date["month"] = df_date["date"].dt.month
df_date["day"] = df_date["date"].dt.day

print(df_date)
```

```python
# Task 12

start_date = "2026-01-15"
end_date = "2026-03-01"

filtered_dates = df_date[
    (df_date["date"] >= start_date) &
    (df_date["date"] <= end_date)
]

print(filtered_dates)
```

```python
# Task 13

data = {
    "start_date": ["2026-01-01", "2026-02-10", "2026-03-05"],
    "end_date": ["2026-01-10", "2026-02-20", "2026-03-15"]
}

df_days = pd.DataFrame(data)

df_days["start_date"] = pd.to_datetime(df_days["start_date"])
df_days["end_date"] = pd.to_datetime(df_days["end_date"])

df_days["difference_days"] = (
    df_days["end_date"] - df_days["start_date"]
).dt.days

print(df_days)
```
