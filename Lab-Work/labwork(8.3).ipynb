```python
# Task 1.1

import numpy as np

ages = np.array([25, 30, 18, 35, 22, 40, 17])

adults = ages[ages >= 18]

print("Adults:", adults)
```

```python
# Task 1.2

ages = np.array([25, 30, 18, 35, 22, 40, 17])

young_adults = ages[(ages > 20) & (ages < 35)]

print("Young Adults:", young_adults)
```

```python
# Task 1.3

names = np.array(["Alice", "Bob", "Charlie", "David", "Eve"])

select = names[
    np.char.startswith(names, "A") |
    np.char.startswith(names, "E")
]

print(select)
```

```python
# Task 2.1

temperatures = np.array([28, 32, 25, 35, 30, 22, 38])

is_hot = temperatures > 30

hot_temperatures = temperatures[is_hot]

print("Mask:", is_hot)
print("Hot temperatures:", hot_temperatures)
```

```python
# Task 2.2

temperatures = np.array([28, 32, 25, 35, 30, 22, 38])

is_hot = temperatures > 30

temperatures[is_hot] = 31

print("Mask:", is_hot)
print("Updated temperatures:", temperatures)
```

```python
# Task 2.3

scores = np.array([65, 80, 72, 90, 55, 78])

grades = np.where(scores >= 75, "Pass", "Fail")

print("Grades:", grades)
```

```python
# Task 4.1

prices = np.array([10, 20, 30])

prices_inc = prices + 5

print(prices_inc)
```

```python
# Task 4.2

sales = np.array([[10, 20],
                  [15, 25],
                  [12, 18]])

bonuses = np.array([2, 3, 1])

broadcasting = sales + bonuses.reshape(3, 1)

print(broadcasting)
```

```python
# Task 4.3

sales = np.array([[10, 20],
                  [15, 25],
                  [12, 18]])

tax_rates = np.array([0.1, 0.05])

broadcasting = sales * tax_rates.reshape(1, 2)

print(broadcasting)
```
