```python
# Task 1

import numpy as np

arr1 = np.array([
    [1, 5, 7],
    [4, 9, 2],
    [6, 10, 14]
])

arr2 = np.array([
    [11, 15, 12],
    [14, 19, 21],
    [16, 15, 17]
])

print("Addition:\n", arr1 + arr2)
print("\nSubtraction:\n", arr1 - arr2)
print("\nMultiplication:\n", arr1 * arr2)
print("\nDivision:\n", arr1 / arr2)
```

```python
# Task 2

arr = np.random.randint(1,26, (3,3))

print(arr)

print("Square Root:\n", np.sqrt(arr))      
print("\nLogarithm (base e):\n", np.log(arr)) 
print("\nExponential:\n", np.exp(arr))  
```

```python
# Task 3

arr = np.array([1, 5, 7, 9, 11])

print("Sine:", np.sin(arr))               
print("Cosine:", np.cos(arr))             
print("Tangent:", np.tan(arr)) 
```

```python
# Task 4

arr = np.array([2, 4, 6, 9, 13])

normalized_arr = (arr - np.mean(arr)) / np.std(arr)

print("Normalized array:", normalized_arr)
```

```python
# Task 5

def euclidean_distance(x1, y1, x2, y2):
    distance = np.sqrt(
        np.pow(x2 - x1, 2) + np.pow(y2 - y1, 2)
    )

    return distance


result = euclidean_distance(4, 9, 8, 12)

print("Distance:", result)
```

```python
# Task 6

arr1 = np.array([5, 7, 9])
arr2 = np.array([3, 5, 8])

arr1_2d = arr1.reshape(1, 3)
arr2_2d = arr2.reshape(1, 3)

horizontal = np.concatenate((arr1_2d, arr2_2d), axis=1)
vertical = np.concatenate((arr1_2d, arr2_2d), axis=0)

print("Horizontal:\n", horizontal)
print("\nVertical:\n", vertical)
```

```python
# Task 7

arr1 = np.array([[1, 3, 5],
                 [4, 7, 9]])

arr2 = np.array([[12, 13, 15],
                 [2, 17, 19]])

arr3 = np.array([[16, 21, 25],
                 [24, 27, 29]])

vstack = np.vstack((arr1, arr2, arr3))

print("Vertical stack:\n", vstack)
print("\nVertical split:\n", np.vsplit(vstack, 3))

hstack = np.hstack((arr1, arr2, arr3))

print("Horizontal stack:\n", hstack)
print("\nHorizontal split:\n", np.hsplit(hstack, 3))
```

```python
# Task 8

arr = np.array([[1, 3, 6, 8],
                [4, 7, 9, 11]])

print("split:\n", np.hsplit(arr, 4))
```

```python
# Task 9

arr1 = np.array([
    [1, 2],
    [3, 4]
])

arr2 = np.array([
    [5, 6],
    [7, 8]
])

concatenated = np.concatenate((arr1, arr2))
stacked = np.stack((arr1, arr2))
horizontal = np.hstack((arr1, arr2))

print("Concatenate:\n", concatenated)
print("\nStack:\n", stacked)
print("\nHorizontal stack:\n", horizontal)
```

```python
# Task 10

arr = np.array([
    [[1, 3], [5, 7]],
    [[9, 11], [13, 15]]
])

flattened_arr = arr.reshape(-1)

print("\nFlattened array:\n", flattened_arr)

original_shape = flattened_arr.reshape(2, 2, 2)

print("\nReshaped to original shape:\n", original_shape)
```

```python
# Task 11

arr = np.array([3, 5, 8, 11, 13, 17])

maximum = np.max(arr)
minimum = np.min(arr)

maximum_index = np.where(arr == maximum)
minimum_index = np.where(arr == minimum)

print("Index of maximum:", maximum_index[0])
print("Index of minimum:", minimum_index[0])
```

```python
# Task 12

arr = np.array([
    [8, 3, 6, 1],
    [4, 11, 9, 7]
])

print("Original array:\n", arr)

print("\nSorted by columns:\n", np.sort(arr, axis=0))

print("\nSorted by rows:\n", np.sort(arr, axis=1))
```

```python
# Task 13

arr = np.random.randint(1, 31, (2, 4))

print("Array:\n", arr)
print("Mean:", np.mean(arr))    
print("Median:", np.median(arr)) 
print("Standard Deviation:", np.std(arr)) 
print("Variance:", np.var(arr))
```

```python
# Task 14

arr = np.array([15, 13, 11, 18, 21, 5, 7, 9, 17])

positions = np.where(arr > 10)

print("Indexes:", positions[0])
print("Numbers greater than 10:", arr[positions])
```

```python
# Task 15

arr = np.random.randint(-10, 21, (4, 4))

positive = np.sum(arr > 0)
negative = np.sum(arr < 0)
zero = np.sum(arr == 0)

print("Array:\n", arr)
print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```
