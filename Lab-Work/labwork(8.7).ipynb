```python
# Task 1

import pandas as pd

data = pd.DataFrame({
    "Month": ["January", "January", "February", "February", "March", "March"],
    "Product": ["Camera", "Watch", "Camera", "Watch", "Camera", "Watch"],
    "Sales": [20, 15, 25, 18, 30, 22]
})

print(data)

pivot_data = data.pivot(index = "Month", columns = "Product", values = "Sales")

print(pivot_data)
```

```python
# Task 2

data = pd.DataFrame({
    "Student": ["Alexander", "Alexander", "Cillian", "Cillian", "Charles", "Charles"],
    "Subject": ["Math", "English", "Math", "English", "Math", "English"],
    "Marks": [80, 75, 85, 78, 90, 82]
})

pivot_students = data.pivot(index = "Student", columns = "Subject", values = "Marks")

print(pivot_students)
```

```python
# Task 3

data = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Project": ["A", "B", "A", "B", "A", "B"],
    "Hours": [40, 35, 30, 25, 45, 38]
})

pivot_hours = data.pivot_table(index = "Department", columns = "Project", values = "Hours")

print(pivot_hours)
```

```python
# Task 4

data = pd.DataFrame({
    "Region": ["North", "North", "South", "South", "East", "East"],
    "Fruit": ["Apple", "Banana", "Apple", "Banana", "Apple", "Banana"],
    "Price": [100, 60, 110, 65, 105, 70]
})

pivot_prices = data.pivot_table(index = "Region", columns = "Fruit", values = "Price", aggfunc = "mean")

print(pivot_prices)
```

```python
# Task 5

data = pd.DataFrame({
    "City": ["Delhi", "Delhi", "Mumbai", "Mumbai", "Pune", "Pune"],
    "Carrier": ["A", "B", "A", "B", "A", "B"],
    "Delivery_Time": [3, 4, 2, 3, 4, 5]
})

pivot_delivery = data.pivot(index = "City",columns = "Carrier",values = "Delivery_Time")

print(pivot_delivery)
```

```python
# Task 7

data = pd.DataFrame({
    "emp_id": [101, 102, 103],
    "emp_name": ["Alexander", "Cillian", "Charles"],
    "dept": ["IT", "HR", "Sales"]
})

data = data.rename(columns={
    "emp_id": "Employee ID",
    "emp_name": "Employee Name",
    "dept": "Department"
})

print(data)
```

```python
# Task 9

inventory = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Speakers"],
    "Quantity": [10, 20, 15]
}, index=["P1", "P2", "P3"])

inventory = inventory.reset_index()

print(inventory)
```

```python
# Task 10

league = pd.DataFrame({
    "Team": ["Team A", "Team B", "Team C"],
    "Points": [25, 30, 20],
    "Wins": [8, 10, 6],
    "Losses": [2, 1, 4]
})

league = league[["Team", "Wins", "Losses", "Points"]]

print(league)
```

```python
# Task 11

sales_data = pd.DataFrame({
    "Region": ["North", "South", "North", "South", "East"],
    "Sales": [1000, 1500, 1200, 1800, 2000]
})

region_sales = sales_data.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
# Task 12

student_data = pd.DataFrame({
    "Class": ["A", "A", "B", "B", "C"],
    "Marks": [80, 90, 70, 85, 88]
})

average_marks = student_data.groupby("Class")["Marks"].mean()

print(average_marks)
```

```python
# Task 13

sales_data = pd.DataFrame({
    "Region": ["North", "North", "South", "South"],
    "Sales": [1000, 1500, 1200, 1800]
})

sales_data["Region_Total"] = (sales_data.groupby("Region")["Sales"].transform("sum"))

sales_data["Sales_Percentage"] = (sales_data["Sales"] / sales_data["Region_Total"]) * 100

print(sales_data)
```

```python
# Task 14

orders = pd.DataFrame({
    "customer_id": [101, 101, 102, 102, 103],
    "Order_Amount": [500, 1200, 800, 1500, 900]
})

maximum_order = (orders.groupby("customer_id")["Order_Amount"].max())

print(maximum_order)
```

```python
# Task 15

employee_data = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [50000, 60000, 40000, 50000, 45000, 55000]
})

employee_data["Standardized_Salary"] = (employee_data.groupby("Department")["Salary"].transform(lambda x: (x - x.mean()) / x.std()))

print(employee_data)
```
