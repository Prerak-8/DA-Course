```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

class SalesDataAnalyzer:
    def __init__(self):
        self.data = None
        self.figure = None

    def __del__(self):
        pass

    def load_data(self):
        path_csv = input("Enter the path of the dataset (CSV file): ")

        try:
            self.data = pd.read_csv(path_csv)
        except FileNotFoundError:
            print("File does not exist. Try again.\n")
        else:
            print("Dataset loaded successfully!\n")    

    def explore_data_first_5_rows(self):
        
        if self.data is None:
            print("Load a dataset first.\n")
            return

        print(self.data.head(), "\n")

    def explore_data_last_5_rows(self):
            
        if self.data is None:
            print("Load a dataset first.\n")
            return

        print(self.data.tail(), "\n")

    def explore_data_col_names(self):
                
        if self.data is None:
            print("Load a dataset first.\n")
            return

        print(self.data.columns, "\n")

    def explore_data_dtypes(self):
                    
        if self.data is None:
            print("Load a dataset first.\n")
            return

        print(self.data.dtypes, "\n")

    def explore_data_info(self):
                        
        if self.data is None:
            print("Load a dataset first.\n")
            return

        self.data.info()
        print()

    def numpy_array_index_slice(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        arr = np.array(self.data)

        print("First row:")
        print(arr[0])

        print("\nFirst 7 rows:")
        print(arr[0:7], "\n")

    def mathematical_operation(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        select_column = input("Type a column to perform numerical operations on: ")

        if select_column not in self.data.columns:
            print("Column not found.\n")
            return

        print("\n1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division\n")

        choice_math = int(input("Enter your choice: "))
        value = float(input("Enter a number: "))

        match choice_math:
            case 1:
                print(self.data[select_column] + value, "\n")

            case 2:
                print(self.data[select_column] - value, "\n")

            case 3:
                print(self.data[select_column] * value, "\n")

            case 4:
                print(self.data[select_column] / value, "\n")

            case _:
                print("Invalid choice.\n")

    def combine_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        path_csv = input("Enter the path of CSV file to combine: ")

        try:
            second_data = pd.read_csv(path_csv)

        except FileNotFoundError:
            print("File does not exist.\n")
            return

        self.data = pd.concat([self.data, second_data])

        print("DataFrames combined successfully!\n")
        print(self.data.head(), "\n")

    def split_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        split_column = input("Enter column to split data by: ")

        if split_column not in self.data.columns:
            print("Column not found.\n")
            return

        split_value = input("Enter value to split by: ")

        first_data = self.data[self.data[split_column].astype(str) == split_value]
        second_data = self.data[self.data[split_column].astype(str) != split_value]

        print("\nFirst DataFrame:")
        print(first_data.head(), "\n")

        print("Second DataFrame:")
        print(second_data.head(), "\n")

    def search_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        search_column = input("Enter column to search in: ")

        if search_column not in self.data.columns:
            print("Column not found.\n")
            return

        search_value = input("Enter value to search: ")

        result = self.data[self.data[search_column].astype(str) == search_value]

        print("\nSearch Result:")
        print(result, "\n")

    def sort_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        sort_column = input("Enter column to sort by: ")

        if sort_column not in self.data.columns:
            print("Column not found.\n")
            return

        print(self.data.sort_values(by = sort_column), "\n")

    def filter_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        filter_column = input("Enter column to filter by: ")

        if filter_column not in self.data.columns:
            print("Column not found.\n")
            return

        filter_value = input("Enter value to filter: ")

        filtered_data = self.data[self.data[filter_column].astype(str) == filter_value]

        print(filtered_data, "\n")

    def create_pivot_table(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        index_column = input("Enter column for rows: ")
        value_column = input("Enter column for values: ")

        if index_column not in self.data.columns or value_column not in self.data.columns:
            print("Column not found.\n")
            return

        pivot = pd.pivot_table(self.data, index = index_column, values = value_column, aggfunc = "sum")

        print("\nPivot Table:")
        print(pivot, "\n")

    def rename_column(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        old_name = input("Enter column name to rename: ")

        if old_name not in self.data.columns:
            print("Column not found.\n")
            return

        new_name = input("Enter new column name: ")

        self.data = self.data.rename(columns={old_name: new_name})

        print("Column renamed successfully!\n")

    def reindex_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        print("Current columns:")
        print(self.data.columns)

        columns = input("Enter column names in new order separated by spaces: ").split()

        self.data = self.data.reindex(columns = columns)

        print(self.data.head(), "\n")

    def groupby_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        group_column = input("Enter column to group by: ")
        value_column = input("Enter numerical column: ")

        if group_column not in self.data.columns or value_column not in self.data.columns:
            print("Column not found.\n")
            return

        result = self.data.groupby(group_column)[value_column].sum()

        print("\nGrouped Data:")
        print(result, "\n")

    def transform_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        transform_column = input("Enter column to transform: ")
        value_column = input("Enter numerical column: ")

        if transform_column not in self.data.columns or value_column not in self.data.columns:
            print("Column not found.\n")
            return

        result = self.data.groupby(transform_column)[value_column].transform("mean")

        print("\nTransformed Data:")
        print(result, "\n")

    def change_data_type(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        column = input("Enter column name: ")

        if column not in self.data.columns:
            print("Column not found.\n")
            return

        data_type = input("Enter new data type (int, float or str): ")

        self.data[column] = self.data[column].astype(data_type)

        print("Data type changed successfully!\n")

    def display_missing_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        print(self.data.isnull(), "\n")

    def fill_missing_mean(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        column = input("Enter numerical column: ")

        if column not in self.data.columns:
            print("Column not found.\n")
            return

        mean_value = self.data[column].mean()

        self.data[column] = self.data[column].fillna(mean_value)

        print("Missing values filled with mean successfully!\n")

    def drop_missing_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        self.data = self.data.dropna()

        print("Rows with missing values dropped successfully!\n")

    def replace_missing_data(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        column = input("Enter column containing missing values: ")

        if column not in self.data.columns:
            print("Column not found.\n")
            return

        value = input("Enter replacement value: ")

        self.data[column] = self.data[column].fillna(value)

        print("Missing values replaced successfully!\n")

    def descriptive_statistics(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        print(self.data.describe(), "\n")

    def aggregate_functions(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        column = input("Enter numerical column: ")

        if column not in self.data.columns:
            print("Column not found.\n")
            return

        print("\nSum:", self.data[column].sum())
        print("Mean:", self.data[column].mean())
        print("Count:", self.data[column].count(), "\n")

    def statistical_analysis(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        column = input("Enter numerical column: ")

        if column not in self.data.columns:
            print("Column not found.\n")
            return

        print("\nStandard Deviation:", self.data[column].std())
        print("Variance:", self.data[column].var())
        print("28th Percentile:", self.data[column].quantile(0.28))
        print("72nd Percentile:", self.data[column].quantile(0.72), "\n")

    def bar_plot(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        x_column = input("Enter category column: ")
        y_column = input("Enter numerical column: ")

        chart_data = self.data.groupby(x_column)[y_column].sum()

        self.figure = plt.figure()

        plt.bar(chart_data.index, chart_data.values)
        plt.title("Bar Chart")
        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.xticks(rotation=45)
        plt.show()

    def line_plot(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        x_column = input("Enter X-axis column: ")
        y_column = input("Enter Y-axis column: ")

        self.figure = plt.figure()

        plt.plot(self.data[x_column], self.data[y_column])
        plt.title("Line Chart")
        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.show()

    def scatter_plot(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        x_column = input("Enter X-axis column: ")
        y_column = input("Enter Y-axis column: ")

        self.figure = plt.figure()

        plt.scatter(self.data[x_column], self.data[y_column])
        plt.title("Scatter Plot")
        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.show()

    def pie_chart(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        category_column = input("Enter category column: ")
        value_column = input("Enter numerical column: ")

        self.figure = plt.figure()

        pie_data = self.data.groupby(category_column)[value_column].sum()

        plt.pie(pie_data.values, labels = pie_data.index)
        plt.title("Pie Chart")
        plt.show()

    def histogram(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        column = input("Enter numerical column: ")

        self.figure = plt.figure()

        plt.hist(self.data[column])
        plt.title("Histogram")
        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.show()

    def stack_plot(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        x_column = input("Enter X-axis column: ")
        first_column = input("Enter first numerical column: ")
        second_column = input("Enter second numerical column: ")

        self.figure = plt.figure()

        plt.stackplot(self.data[x_column], self.data[first_column], self.data[second_column], labels = [first_column, second_column])
        plt.title("Stack Plot")
        plt.xlabel(x_column)
        plt.legend()
        plt.show()

    def subplot_chart(self):
    
        if self.data is None:
            print("Load a dataset first.\n")
            return

        x_column = input("Enter X-axis column: ")
        y1_column = input("Enter first Y-axis column: ")
        y2_column = input("Enter second Y-axis column: ")

        self.figure, axes = plt.subplots(1, 2)

        axes[0].plot(self.data[x_column], self.data[y1_column])
        axes[0].set_title("First Plot")

        axes[1].plot(self.data[x_column], self.data[y2_column])
        axes[1].set_title("Second Plot")

        plt.show()

    def seaborn_box_plot(self):

        if self.data is None:
            print("Load a dataset first.\n")
            return

        x_column = input("Enter category column: ")
        y_column = input("Enter numerical column: ")

        sns.set_theme()

        self.figure = plt.figure()

        sns.boxplot(data = self.data, x = x_column, y = y_column)
        plt.title("Seaborn Box Plot")
        plt.show()

    def save_visualization(self):

        if self.figure is None:
            print("Create a visualization first.\n")
            return

        file_name = input("Enter file name to save visualization as (like this example: chart.png): ")

        self.figure.savefig(file_name)

        print(f"Visualization saved as {file_name} successfully!\n")

    

analyzer = SalesDataAnalyzer()

while True:
    print("========== Data Analysis & Visualization Program ==========\n")
    print("Please select an option:")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")
    print("===========================================================\n")

    choice = int(input("Enter your choice: "))
    print()

    match choice:
        case 1:
            print("~~ Load DataSet ~~")
            analyzer.load_data()

        case 2:
            while True:
                print("~~ Explore Data ~~")
                print("1. Display the first 5 rows")
                print("2. Display the last 5 rows")
                print("3. Display column names")
                print("4. Display data types")
                print("5. Display basic info")
                print("6. Back to Main Menu.\n")

                choice_exp_data = int(input("Enter your choice: "))
                print()

                match choice_exp_data:
                    case 1:
                        analyzer.explore_data_first_5_rows()

                    case 2:
                        analyzer.explore_data_last_5_rows()

                    case 3:
                        analyzer.explore_data_col_names()

                    case 4:
                        analyzer.explore_data_dtypes()

                    case 5:
                        analyzer.explore_data_info()

                    case 6:
                        break

                    case _:
                        print("Invalid choice entered. Please enter a valid choice.\n")

        case 3:
            while True:
                print("~~ Perform DataFrame Operations ~~")
                print("1. NumPy Array Indexing and Slicing")
                print("2. Mathematical Operations")
                print("3. Combine DataFrames")
                print("4. Split DataFrame")
                print("5. Search, Sort and Filter")
                print("6. Create Pivot Table")
                print("7. Reindex / Rename Labels")
                print("8. GroupBy and Transform")
                print("9. Change data type")
                print("10. Back to Main Menu.\n")
            
                choice_perf_data = int(input("Enter your choice: "))
                print()
            
                match choice_perf_data:
                    case 1:
                        analyzer.numpy_array_index_slice()
            
                    case 2:
                        analyzer.mathematical_operation()
            
                    case 3:
                        analyzer.combine_data()
            
                    case 4:
                        analyzer.split_data()
            
                    case 5:
                        analyzer.search_data()
                        analyzer.sort_data()
                        analyzer.filter_data()
            
                    case 6:
                        analyzer.create_pivot_table()

                    case 7:
                        analyzer.reindex_data()
                        analyzer.rename_column()

                    case 8:
                        analyzer.groupby_data()
                        analyzer.transform_data()

                    case 9:
                        analyzer.change_data_type()

                    case 10:
                        break
            
                    case _:
                        print("Invalid choice entered. Please enter a valid choice.\n")

        case 4:
            while True:
                print("~~ Handle Missing Data ~~")
                print("1. Display missing values")
                print("2. Fill missing values with mean")
                print("3. Drop rows with missing values")
                print("4. Replace missing values with a specific value")
                print("5. Back to Main Menu.\n")

                choice_missing = int(input("Enter your choice: "))
                print()

                match choice_missing:
                    case 1:
                        analyzer.display_missing_data()

                    case 2:
                        analyzer.fill_missing_mean()

                    case 3:
                        analyzer.drop_missing_data()

                    case 4:
                        analyzer.replace_missing_data()

                    case 5:
                        break

                    case _:
                        print("Invalid choice entered. Please enter a valid choice.\n")

        case 5:
            while True:
                print("~~ Generate Descriptive Statistics ~~")
                print("1. Display Descriptive Statistics")
                print("2. Aggregate Functions")
                print("3. Statistical Analysis")
                print("4. Back to Main Menu.\n")

                choice_stats = int(input("Enter your choice: "))
                print()

                match choice_stats:
                    case 1:
                        analyzer.descriptive_statistics()

                    case 2:
                        analyzer.aggregate_functions()

                    case 3:
                        analyzer.statistical_analysis()

                    case 4:
                        break

                    case _:
                        print("Invalid choice entered. Please enter a valid choice.\n")

        case 6:
            while True:
                print("~~ Data Visualization ~~")
                print("1. Bar Plot")
                print("2. Line Plot")
                print("3. Scatter Plot")
                print("4. Pie Chart")
                print("5. Histogram")
                print("6. Stack Plot")
                print("7. Subplots")
                print("8. Seaborn Box Plot")
                print("9. Back to Main Menu.\n")

                choice_visual = int(input("Enter your choice: "))
                print()

                match choice_visual:
                    case 1:
                        analyzer.bar_plot()

                    case 2:
                        analyzer.line_plot()

                    case 3:
                        analyzer.scatter_plot()

                    case 4:
                        analyzer.pie_chart()

                    case 5:
                        analyzer.histogram()

                    case 6:
                        analyzer.stack_plot()

                    case 7:
                        analyzer.subplot_chart()

                    case 8:
                        analyzer.seaborn_box_plot()

                    case 9:
                        break

                    case _:
                        print("Invalid choice entered. Please enter a valid choice.\n")

        case 7:
            print("~~ Save Visualization ~~")
            analyzer.save_visualization()

        case 8:
            break

        case _:
            print("Invalid choice entered. Please enter a valid choice.\n")

print("Thank you for using Data Analysis & Visualization Program!\nHave a nice day.\nGoodbye.")
```
