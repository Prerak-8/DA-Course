```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print("============== Global Happiness Report Analysis ==============\n")

data = pd.read_csv("data-2019.csv")
```

```python
print(data.head(), "\n")
print(data.tail(), "\n")
data.info()
print()
print(data.describe(), "\n")
print(data.columns, "\n")
```

```python
highest_hap_score = data["Score 2019"].max()
highest_country = data[data["Score 2019"] == highest_hap_score]["Country"]

print("Highest happiness score:", highest_hap_score)
print("Country with highest happiness score:")
print(highest_country, "\n")

lowest_hap_score = data["Score 2019"].min()
lowest_country = data[data["Score 2019"] == lowest_hap_score]["Country"]

print("Lowest happiness score:", lowest_hap_score)
print("Country with lowest happiness score:")
print(lowest_country, "\n")

average_score = data["Score 2019"].mean()
print("Average happiness score in 2019:", round(average_score, 3), "\n")
```

```python
print("Missing values:")
print(data.isnull().sum(), "\n")
print("Duplicate rows:", data.duplicated().sum(), "\n")
```

```python
top_10 = data.sort_values(by="Score 2019", ascending=False).head(10)

plt.bar(top_10["Country"], top_10["Score 2019"])
plt.title("Top 10 Happiest Countries")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(data["GDP 2019"], data["Score 2019"])
plt.title("Happiness Score vs GDP")
plt.xlabel("GDP")
plt.ylabel("Happiness Score")
plt.show()
```

```python
life_data = data.sort_values(by="Life Expectancy 2019")

plt.plot(life_data["Life Expectancy 2019"], life_data["Score 2019"])
plt.title("Happiness Score vs Life Expectancy")
plt.xlabel("Life Expectancy")
plt.ylabel("Happiness Score")
plt.show()
```

```python
region_happiness = data.groupby("Region")["Score 2019"].mean()

plt.bar(region_happiness.index, region_happiness.values)
plt.title("Average Happiness Score by Region")
plt.xlabel("Region")
plt.ylabel("Average Happiness Score")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.hist(data["Score 2019"])
plt.title("Distribution of Happiness Scores")
plt.xlabel("Happiness Score")
plt.ylabel("Frequency")
plt.show()
```

```python
numeric_data = data.drop(columns=["Country", "Region"])
correlation = numeric_data.corr()

sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
